```python
import pandas as pd

# Load the cleaned dataset
df = pd.read_csv("cleaned_delayed_flights.csv")
print("✅ Cleaned dataset loaded.")
```

```python
df.columns
```

```python
# Example split: January as reference, July as current (seasonal drift)
reference_data = df[df["Month"] == 9].copy()
current_data = df[df["Month"] == 10].copy()
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split

# Select features and target
target = "Arrival Delay"
features = ['Month', 'Day Of Month', 'Day Of Week', 'Departure Delay', 'Distance', 'Diverted']

X_ref = reference_data[features]
y_ref = reference_data[target]

# Split ref data for training/testing
X_train, X_test, y_train, y_test = train_test_split(X_ref, y_ref, test_size=0.3, random_state=42)

# Train model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Predict on reference and current data
reference_data["model_prediction"] = model.predict(reference_data[features])
current_data["model_prediction"] = model.predict(current_data[features])
```

```python
import evidently

# List top-level modules, classes, functions
print(dir(evidently.metrics))
```

```python
from evidently.report import Report
from evidently.metrics import (
    ClassificationQualityMetric,
    AccuracyMetric,
    PrecisionMetric,
    RecallMetric,
    F1Metric
)
from evidently import ColumnMapping

# Create the report
report = Report(metrics=[
    ClassificationQualityMetric(),
    AccuracyMetric(),
    PrecisionMetric(),
    RecallMetric(),
    F1Metric()
])

# Define the column mapping using ColumnMapping class
column_mapping = ColumnMapping(
    target="Delayed",
    prediction="model_prediction"
)

# Run the report
report.run(
    reference_data=reference_data,
    current_data=current_data,
    column_mapping=column_mapping
)

# Save the report
report.save_html("model_quality_report.html")
```
