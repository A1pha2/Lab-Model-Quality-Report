```python
# Check for columns with null values
null_columns = df.columns[df.isnull().any()]
print("Columns with null values:", null_columns)
```

```python
df = df.drop('Unnamed: 0', axis=1)
col_names = ['Year','Month', 'Day Of Month','Day Of Week','Departure Time', 'Scheduled Departure Time','Arrival Time', 'Scheduled Arrival Time', 'Unique Carrier Code', 
             'Flight Number','Tail Number', 'Actual Elapsed Time','Estimated Elapsed Time','Air Time', 'Arrival Delay','Departure Delay','Origin', 'Destination',
             'Distance','Taxi In','Taxi Out', 'Cancelled', 'Cancellation Code','Diverted','Carrier Delay','Weather Delay', 'NAS Delay','Security Delay','Late Aircraft Delay']

# Fix column names, drop duplicates, show all columns
df.columns = col_names
df = df.drop_duplicates()
pd.set_option('display.max_columns', None)

df
```

```python
invalid_years = df[df['Year'] != 2008]
if not invalid_years.empty:
    print("Invalid Years:")
    print(invalid_years['Year'])

# Check month and print the rows where the month is not in the range 1-12
invalid_months = df[~df['Month'].between(1, 12)]
if not invalid_months.empty:
    print("Invalid Months:")
    print(invalid_months['Month'])

# Check day of the month and print the rows where the day is not in the range 1-31
invalid_month_day = df[~df['Day Of Month'].between(1, 31)]
if not invalid_month_day.empty:
    print("Invalid Month Day:")
    print(invalid_months['Day Of Month'])

# Check day of the week and print the rows where the day is not in the range 1-7
invalid_day = df[~df['Day Of Week'].between(1, 7)]
if not invalid_month_day.empty:
    print("Invalid Week Day:")
    print(invalid_months['Day Of Week'])
```

```python
null_df_sums = df.isna().sum()
null_df_sums
```

```python
'''
    There are 5 missing values for Tail Number, since these are needed to track the plane, 
    we can claim the data for that missing row is invalid
'''
df.dropna(subset=['Tail Number'], inplace=True)
```

```python
''' 
Given:
    We already know the Cancelled and Cancellation Code columns are nonempty because the sum of their .isna()
    in the code cell above was equal to 0.

    If flight was cancelled (cancelled = 1), Taxi In, Taxi Out = 0.0
    
'''

cancelled_flights = df[df['Cancelled'] == 1]
cancelled_flight_codes = df[df['Cancellation Code'] != 'N']

print('There are',len(cancelled_flights), 'cancelled flights.')
print('There are',len(cancelled_flight_codes),'cancelled flight codes')

# Print the dataframe of the cancelled flight rows
cancelled_flights
```

```python
'''
    We see that all cancelled flights have NaN for Taxi in, actual elapsed time, and NaN for all delay codes
    Let's and write "0.0" for arrival time, actual elapsed time, air time, arrival delay, taxi in, and taxi out
    for NaN values in cancelled flights.
'''
df['Arrival Time'] = df['Arrival Time'].fillna(value=0.0)
df['Actual Elapsed Time'] = df['Actual Elapsed Time'].fillna(value=0.0)
df['Air Time'] = df['Air Time'].fillna(value=0.0)
df['Arrival Delay'] = df['Arrival Delay'].fillna(value=0.0)
df['Taxi In'] = df['Taxi In'].fillna(value=0.0)
df['Taxi Out'] = df['Taxi Out'].fillna(value=0.0)
```

```python
cause_cols = ["Carrier Delay", "Weather Delay", "NAS Delay", "Security Delay", "Late Aircraft Delay"]

df = df.drop(columns=cause_cols)

print("Dataset shape after dropping rows with NaN delay causes:", df.shape)
```

```python
# Check for columns with null values
null_columns = df.columns[df.isnull().any()]
print("Columns with null values:", null_columns)
```

```python
# Drop rows where flight was cancelled
df = df[df["Cancelled"] != 1]
# Drop the specified columns
columns_to_drop = [
    'Scheduled Departure Time', 'Arrival Time', 'Scheduled Arrival Time', 
    'Unique Carrier Code', 'Flight Number', 'Tail Number', 
    'Actual Elapsed Time', 'Estimated Elapsed Time', 
    'Origin', 'Destination', 'Taxi In', 'Taxi Out', 
    'Cancelled', 'Cancellation Code'
]

df = df.drop(columns=columns_to_drop)

print("Data shape after dropping cancelled rows and columns:", df.shape)
```
